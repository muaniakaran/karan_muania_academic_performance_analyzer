## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_01_dataset.csv")
display(df.head())

```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate rows
print("Dataset shape (rows, columns):", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
subject_columns = ["Python", "Maths", "Physics"]

marks_array = df[subject_columns].to_numpy()

print("NumPy array of subject marks:")
print(marks_array)

print("\nAverage of all subject marks:", np.mean(marks_array))
print("Highest mark:", np.max(marks_array))
print("Lowest mark:", np.min(marks_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
analysis = df.copy()

analysis["Total"] = analysis[subject_columns].sum(axis=1)
analysis["Percentage"] = (analysis["Total"] / 300) * 100
analysis["Average"] = analysis[subject_columns].mean(axis=1)

display(analysis.head())
```

## 6. Your individual project challenge
**Challenge:** Calculate total and percentage; identify topper and students below 40 in any subject; compare subject averages.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
below_40 = analysis[
    (analysis["Python"] < 40) |
    (analysis["Maths"] < 40) |
    (analysis["Physics"] < 40)
]

print("Students scoring below 40 in at least one subject:")
display(below_40[["Student_ID", "Name", "Python", "Maths", "Physics"]])

sorted_students = analysis.sort_values(by="Total", ascending=False)

print("Students sorted by total marks (highest first):")
display(sorted_students[["Student_ID", "Name", "Total", "Percentage"]])
```
